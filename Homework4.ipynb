```python
# Q1 solution:

import spacy

# Load the English model
# (If not installed, run these in a terminal or Colab cell first:)
# !pip install spacy
# !python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")

text = "John enjoys playing football while Mary loves reading books in the library."
doc = nlp(text)

# Remove stopwords and keep only nouns and verbs (lemmatized)
result = [token.lemma_ for token in doc if not token.is_stop and token.pos_ in ("NOUN", "VERB")]

print("Original Text:")
print(text)
print("\nProcessed Output (Lemmatized Nouns & Verbs):")
print(result)


# Q2 solution:

import spacy

# Load the English model
nlp = spacy.load("en_core_web_sm")

text = "Chris met Alex at Apple headquarters in California. He told him about the new iPhone launch."
doc = nlp(text)

print("Named Entities Detected:")
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_}")

# Detect pronouns that may cause ambiguity
pronouns = [tok.text for tok in doc if tok.lower_ in ("he", "she", "they", "him", "her", "them")]
if pronouns:
    print("\nWarning: Possible pronoun ambiguity detected!")
    print("Pronouns found:", pronouns)
else:
    print("\nNo ambiguous pronouns found.")
```
